```python
df = pd.read_csv('/content/population_by_country_2020.csv',encoding="latin1")
```

```python
max_population = df['Population (2020)'].max()
print(f"The maximum population is: {max_population}")
```

```python
min_population = df['Population (2020)'].min()
print(f"The minimum population is: {min_population}")
```

```python
df = df.fillna(0)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.size
```

```python
top_10_countries_with_max = df.sort_values(by='Population (2020)', ascending=False).head(10)
print(top_10_countries_with_max)
```

```python
top_10_countries_with_min = df.sort_values(by='Population (2020)', ascending=True).head(10)
print(top_10_countries_with_min)
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_10_countries_with_max['Country (or dependency)'], top_10_countries_with_max['Population (2020)'])
plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Population by Country')
plt.xticks(rotation=90)
plt.savefig("top_10_countries_with_max_bar.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Country (or dependency)', y='Population (2020)',
            data=top_10_countries_with_max, palette='viridis')
plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Population by Country')
plt.xticks(rotation=90)
plt.savefig("top_10_countries_with_max_barbysns.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_10_countries_with_min['Country (or dependency)'], top_10_countries_with_min['Population (2020)'])
plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Population by Country')
plt.xticks(rotation=90)
plt.savefig("top_10_countries_with_min_bar.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.pie(top_10_countries_with_max['Population (2020)'],
        labels=top_10_countries_with_max['Country (or dependency)'],
        autopct='%1.1f%%')
plt.title('Population by Country in 2020')
plt.savefig("top_10_countries_with_max_pie.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.pie(top_10_countries_with_min['Population (2020)'],
        labels=top_10_countries_with_min['Country (or dependency)'],
        autopct='%1.1f%%')
plt.title('Population by Country in 2020')
plt.savefig("top_10_countries_with_min_pie.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(top_10_countries_with_max['Population (2020)'], top_10_countries_with_max['Yearly Change'], marker='o')
plt.xlabel('Population')
plt.ylabel('Yearly Change')
plt.title('Population vs Yearly Change')
plt.savefig("top_10_countries_with_max_plot.png", dpi=150)
plt.show()
```
